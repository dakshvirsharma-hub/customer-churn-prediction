```python
X= df.drop(columns=['Churn Label','Zip Code','CustomerID','Country','State','City','Lat Long','Latitude','Longitude','Churn Score','CLTV','Churn Reason','Churn Label','Churn Value','Count'])
y = df['Churn Value']
```

```python
 X.select_dtypes(include='object').columns
```

```python
X["Total Charges"].map(type).value_counts()
```

```python
X['Total Charges'] = pd.to_numeric(X['Total Charges'],errors="coerce")
X["Total Charges"].map(type).value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
label_encoders = {}

for i in X.columns:
    if X[i].nunique() == 2 and X[i].dtype == "object":
        le = LabelEncoder()
        X[i] = le.fit_transform(X[i])
        label_encoders[i] = le






multi_col = [col for col in X.select_dtypes(include="object").columns
             if X[col].nunique() > 2]
ohe = OneHotEncoder(drop='first', sparse_output=False)

encoded = ohe.fit_transform(X[multi_col])

encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(multi_col),
    index=X.index
)

X = X.drop(columns=multi_col)
X = pd.concat([X, encoded_df], axis=1)
```

```python
X['Total Charges'].isnull().sum()
```

```python
X['Total Charges'] = X['Total Charges'].fillna(X['Total Charges'].median())
```

```python
X.isnull().sum().sum()
```

```python
from scipy.stats import pearsonr
import pandas as pd

result = []

for col in X.columns:
    r, p = pearsonr(X[col], y)

    result.append({
        "Feature": col,
        "Pearson_r": r,
        "P_value": p
    })

corr_df = pd.DataFrame(result)

corr_df = corr_df.sort_values(
    by="Pearson_r",
    ascending=False
).reset_index(drop=True)

corr_df
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
St = StandardScaler()
x_train_scaler = St.fit_transform(X_train)
x_test_scaler = St.transform(X_test)

```

```python
from sklearn.linear_model import LogisticRegression
le = LogisticRegression()
le.fit(x_train_scaler,y_train)
```

```python
y_pred = le.predict(x_test_scaler)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
print("Accuracy :-", accuracy_score(y_test,y_pred))
print("Precision :-", precision_score(y_test,y_pred))
print("Recall :-", recall_score(y_test,y_pred))
print("f1_score :-",f1_score(y_test,y_pred))

print("\n Confusion matrix")
print(confusion_matrix(y_test,y_pred))

print("\n Classification Report")
print(classification_report(y_test,y_pred))
```

```python
result= []
result.append({
    "Model" : "Logistic regression",
    "Accuracy" : accuracy_score(y_test,y_pred) ,
    "Precision" : precision_score(y_test,y_pred),
    "Recall" : recall_score(y_test,y_pred),
    "F1" : f1_score(y_test,y_pred)

})
compare_df = pd.DataFrame(result)
compare_df
```

```python
param_grid = [
    {
        "penalty": ["l1"],
        "solver": ["liblinear", "saga"]
    },
    {
        "penalty": ["l2"],
        "solver": ["lbfgs", "liblinear", "saga"]
    },
    {
        "penalty": ["elasticnet"],
        "solver": ["saga"],
        "l1_ratio": [0.2, 0.5, 0.8]
    }
]
```

```python
from sklearn.model_selection import GridSearchCV
import numpy as np
grid = GridSearchCV(
    LogisticRegression(max_iter=5000),
    param_grid,
    cv=5,
    scoring="f1",
    error_score=np.nan  # skips invalid combinations
)
```

```python
grid.fit(x_train_scaler,y_train)
```

```python
y_pred_grid = grid.predict(x_test_scaler)
```

```python
result.append({
    "Model": "Logistic Regression (Tuned)",
    "Accuracy": accuracy_score(y_test, y_pred_grid),
    "Precision": precision_score(y_test, y_pred_grid),
    "Recall": recall_score(y_test, y_pred_grid),
    "F1": f1_score(y_test, y_pred_grid)
})

compare_df = pd.DataFrame(result)
compare_df
```

```python
print(grid.best_params_)
print(grid.best_score_)
```
